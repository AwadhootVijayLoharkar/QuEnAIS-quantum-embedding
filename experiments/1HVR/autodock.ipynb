```python
from pdbfixer import PDBFixer
from openmm.app import PDBFile

def clean_protein(input_pdb):
    fixer = PDBFixer(filename=input_pdb)

    # Fix issues
    fixer.findMissingResidues()
    fixer.findMissingAtoms()
    fixer.addMissingAtoms()
    fixer.addMissingHydrogens(pH=7.0)

    # Save cleaned protein
    output = "/home/loharkar/QuEnAIS-quantum-embedding/data/intermediate/cleaned_protein.pdb"
    with open(output, "w") as f:
        PDBFile.writeFile(fixer.topology, fixer.positions, f)

    return output

cleaned_protein = clean_protein("/home/loharkar/QuEnAIS-quantum-embedding/data/raw/1HVR.pdb")
print("Saved:", cleaned_protein)
```

```python
def extract_ligand(pdb_file, ligand_name="XK2"):
    ligand_lines = []
    
    for line in open(pdb_file):
        if line.startswith("HETATM") and ligand_name in line:
            ligand_lines.append(line)

    output = "/home/loharkar/QuEnAIS-quantum-embedding/data/intermediate/ligand.pdb"
    
    with open(output, "w") as f:
        f.writelines(ligand_lines)

    return output


ligand_file = extract_ligand(
    "/home/loharkar/QuEnAIS-quantum-embedding/data/intermediate/cleaned_protein.pdb",
    "XK2"
)

print("Ligand saved:", ligand_file)
```

```python
def remove_ligand(pdb_file, ligand_name="XK2"):
    output = "/home/loharkar/QuEnAIS-quantum-embedding/data/intermediate/protein_only.pdb"
    
    with open(output, "w") as f:
        for line in open(pdb_file):
            # Skip lines that belong to the ligand
            if line.startswith("HETATM") and ligand_name in line:
                continue
            f.write(line)

    return output


protein_only = remove_ligand(
    "/home/loharkar/QuEnAIS-quantum-embedding/data/intermediate/cleaned_protein.pdb",
    "XK2"
)

print("Protein only saved:", protein_only)
```

```python
import numpy as np

def read_pdb_atoms(pdb_file):
    atoms = []

    for line in open(pdb_file):
        if line.startswith(("ATOM", "HETATM")):
            atom = {
                "element": line[76:78].strip(),
                "x": float(line[30:38]),
                "y": float(line[38:46]),
                "z": float(line[46:54]),
                "residue": line[17:20].strip()
            }
            atoms.append(atom)

    return atoms
```

```python
ligand_atoms = read_pdb_atoms(
    "/home/loharkar/QuEnAIS-quantum-embedding/data/intermediate/ligand.pdb"
)

protein_atoms = read_pdb_atoms(
    "/home/loharkar/QuEnAIS-quantum-embedding/data/intermediate/protein_only.pdb"
)

print("Ligand atoms:", len(ligand_atoms))
print("Protein atoms:", len(protein_atoms))
```

```python
def compute_center(atoms):
    coords = np.array([[a["x"], a["y"], a["z"]] for a in atoms])
    return coords.mean(axis=0)

ligand_center = compute_center(ligand_atoms)
print("Ligand center:", ligand_center)
```

```python
from rdkit import Chem
from rdkit.Chem import AllChem

smiles = {
    "L1": "c1ncc[nH]1",
    "L2": "Cc1ncc[nH]1",
    "L3": "c1ccncc1"
}

for name,smi in smiles.items():
    mol = Chem.MolFromSmiles(smi)
    mol = Chem.AddHs(mol)
    AllChem.EmbedMolecule(mol)
    AllChem.UFFOptimizeMolecule(mol)

    Chem.MolToXYZFile(mol, f"{name}.xyz")
```

```python
import numpy as np
from pyscf import gto, scf

def clean_extended_xyz(input_file, output_file="clean.xyz"):
    with open(input_file, 'r') as f:
        lines = f.readlines()
    
    natoms = int(lines[0].strip())
    # Skip the header line (line 1)
    atom_lines = lines[2:2 + natoms]   # the actual atom lines
    
    with open(output_file, 'w') as f:
        f.write(f"{natoms}\n")
        f.write("Cleaned for PySCF\n")
        for line in atom_lines:
            parts = line.split()
            # Take only: element (parts[0]), x, y, z (parts[1:4])
            clean_line = f"{parts[0]}  {parts[1]}  {parts[2]}  {parts[3]}\n"
            f.write(clean_line)

# Use it once
clean_extended_xyz("L1_complex.xyz")

# Now run your calculation on the clean file
mol = gto.Mole()
mol.atom = "clean.xyz"          # or mol.atom = open("clean.xyz").read()
mol.basis = 'sto-3g'
mol.build()

mf = scf.(mol)
E = mf.kernel()
print("Energy:", E)
```

```python
from pyscf import gto, scf, dft
from asf.wrapper import find_from_scf   # preferred

# ... your mol definition with charge=2, spin=0 ...

# First get a good mean-field yourself (much more reliable than ASF's internal one)
mf = scf.RHF(mol)
mf.max_cycle = 150
mf.level_shift = 0.25
mf.damp = 0.35
mf.init_guess = 'huckel'
mf = mf.density_fit()          # helps a lot
mf.kernel()

# If RHF still struggles, fall back to DFT (very common and accepted for embedding workflows)
if not mf.converged:
    print("RHF failed → using DFT")
    mf = dft.RKS(mol)
    mf.xc = 'b3lyp'            # or 'pbe0'
    mf = mf.density_fit()
    mf.kernel()

# Now feed the converged mf to ASF
nel, mo_list, mo_coeff = find_from_scf(
    mf,
    entropy_threshold=0.15,     # 0.1–0.2 is typical; lower = bigger active space
    mp2_max_orb=25,             # limit the MP2 window to speed it up (important for your size)
    # states=1,                 # or [(0,0)] for singlet if needed
)

print(f"Active space: {nel} electrons in {len(mo_list)} orbitals")
print("Active MO indices:", mo_list)
```
